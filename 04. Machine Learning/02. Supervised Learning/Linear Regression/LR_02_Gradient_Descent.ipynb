```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = {
    'x':[6,8,12,14,18],
    'y':[350,775,1150,1395,1675]
}
df = pd.DataFrame(data)
```

```python
x = df['x'].to_numpy()
y = df['y'].to_numpy()
w0=0.0
w1=0.0
lr = 0.01
n=len(x)
for _ in range(25000):
    y_pred = [(w1*i + w0) for i in x]
    j_w0 = np.sum(y_pred-y)/n
    j_w1 = np.sum((y_pred-y)*x)/n
    w0 = w0-lr*j_w0
    w1 = w1-lr*j_w1
print(f'w0 = {w0}\nw1 = {w1}')
```

# R Squared Value

```python
y1_list2 = [y for y in df['y']]
y2_list2 = [(w1*x+w0) for x in df['x']]
avg_y = np.mean(df['y'])

numerator2 = denominator2 = 0

numerator2 = sum((y2-avg_y)**2 for y2 in y2_list2)
denominator2 = sum((y1-avg_y)**2 for y1 in y1_list2)

r_squared_val2 = (numerator2/denominator2).round(2)

print(f'R-Squared Value: {r_squared_val2}')
```
