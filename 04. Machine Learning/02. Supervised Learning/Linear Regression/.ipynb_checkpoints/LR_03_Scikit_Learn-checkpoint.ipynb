# Multiple linear regression

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Libraries for linear regression
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
```

```python
data = {
    'square_feet': [1400, 1600, 1700, 1875, 1100, 1550, 2350, 2450, 1425, 1700],
    'bedrooms':    [3, 3, 4, 4, 2, 3, 4, 5, 3, 3],
    'age_years':   [20, 15, 18, 10, 30, 12, 5, 8, 22, 14],
    'price':       [245000, 312000, 279000, 308000, 199000, 219000, 405000, 324000, 235000, 290000],
}
df = pd.DataFrame(data)
df
```

```python
x = df[['square_feet', 'bedrooms', 'age_years']]
y = df['price']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

model = LinearRegression()

model.fit(x_train, y_train)

predictions = model.predict(x_test)

print(f'Intercept (w0): {model.intercept_}\n')
print(f'Coefficients (w1): {dict(zip(x.columns, model.coef_))}\n')
print(f'R Squared Value: {r2_score(y_test, predictions)}\n')
```
