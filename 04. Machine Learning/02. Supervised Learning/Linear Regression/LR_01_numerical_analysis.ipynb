# Linear Regression:
### y_pred = w0 + w1*x
```
Here,
    w1 = slope
        It measures the change in the mean value of y
        If, 
            w1 = (+)ve -> y increases as x increases
            w1 = (-)ve -> y decreases as x increases
        
    w0 = The y-intercept.
        It is the mean value of y when x equals 0

    
x	y	  xy	  x^2
6	350	  2100	  36
8	775	  6200	  64
12	1150  13800	  144
14	1395  19530	  196
18	1675  30150	  324


Here,

avg_x = (6+8+12+14+18) / 5 = 11.6
avg_y = (350+775+1150+1395+1675) / 5 = 1069
avg_xy = (2100+6200+13800+19530+30150) / 5 = 14356
(avg_x)^2 = 134.56
avg_x^2 = (36+64+144+196+324) / 5 = 152.8

w1 = ((avg_x*avg_y) - avg_xy) / ((avg_x)^2 - avg_x^2)
w0 = avg_y - (m*avg_x)


y_pred = w0 + w1*x

```
    

````python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
data = {
    'x':[6,8,12,14,18],
    'y':[350,775,1150,1395,1675]
}
df = pd.DataFrame(data)
df
````

````python
sns.lmplot(data=df, x='x', y='y')
````

````python
df['xy'] = df['x']*df['y']
df['x^2'] = df['x']**2

print(df, '\n')

avg_x = df['x'].mean()
avg_y = df['y'].mean()
avg_x_sq = df['x^2'].mean()
avg_xy = df['xy'].mean()

print(f'Average x: {avg_x}\nAverage y: {avg_y}\nAverage XY: {avg_xy}\nAverage X^2: {avg_x_sq}\n')

w1 = (((avg_x*avg_y) - avg_xy) / (avg_x**2 - avg_x_sq)).round(2)
w0 = (avg_y - (m*avg_x)).round(2)

print(f'w0 = {w0}\nw1 = {w1}\n')

y_pred = [(w0 + w1*x) for x in df['x']]

print('x = ', *df['x'], end=' ')
print()
print('y_pred = ', *y_pred, end=' ')
````
